```python
#program 1
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder

# Load Heart Disease Dataset
df = pd.read_csv("heart.csv")

# Convert categorical columns into numeric values
encoder = LabelEncoder()

for column in df.columns:
    if df[column].dtype == 'object':
        df[column] = encoder.fit_transform(df[column])

# Select only numerical columns
num_df = df.select_dtypes(include='number')

# Shape of dataset
print("Shape of Dataset:", df.shape)

# Independent and Dependent Features
X = df.drop("HeartDisease", axis=1)
y = df["HeartDisease"]

print("\nIndependent Features:\n", X.columns)
print("\nDependent Feature:\n", y.name)

# Histograms for numerical features
num_df.hist(
    bins=30,
    figsize=(12, 8),
    color='steelblue',
    edgecolor='black'
)

plt.suptitle('Histograms of Numerical Features')
plt.show()

# Box plots for numerical features
num_df.plot(
    kind='box',
    subplots=True,
    layout=(4, 4),
    figsize=(14, 10)
)

plt.suptitle('Box Plots of Numerical Features')
plt.show()

# Outlier Detection using IQR
Q1 = num_df.quantile(0.25)
Q3 = num_df.quantile(0.75)

IQR = Q3 - Q1

outliers_count = (
    (
        (num_df < (Q1 - 1.5 * IQR)) |
        (num_df > (Q3 + 1.5 * IQR))
    ).sum()
)

print("\nOutliers Detection:\n")
print(outliers_count)

# Dataset summary
print("\nDataset Summary:\n")
print(df.describe())
```

```python
# PROGRAM 2: Correlation Matrix & Pair Plot
# Using Heart Disease Dataset

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder

# Load dataset
df = pd.read_csv("heart.csv")

# Convert categorical columns to numeric
encoder = LabelEncoder()

for column in df.columns:
    if df[column].dtype == 'object':
        df[column] = encoder.fit_transform(df[column])

# Correlation Matrix Heatmap
plt.figure(figsize=(12, 10))

sns.heatmap(
    df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()

# Pair Plot
sns.pairplot(
    df.sample(500),
    corner=True,
    diag_kind='kde',
    plot_kws={'alpha': 0.5}
)

plt.suptitle('Pair Plot of Heart Disease Dataset', y=1.02)
plt.show()

# Dataset Shape
print("Shape of Dataset:", df.shape)

# Independent and Dependent Features
X = df.drop("HeartDisease", axis=1)
y = df["HeartDisease"]

print("\nIndependent Features:\n", X.columns)

print("\nDependent Feature:\n", y.name)
```
